# Use the Glances API in your Python code

During the Glances 5 alphas the module is `glances.api_v5`. It becomes `glances.api` with Glances 5.0.0.

## Init the Glances API

By default the API collects on demand: reading a plugin updates it when its last update is older than `[global] refresh` (2 seconds by default).

```python
from glances import api_v5 as api
```

```python
gl = api.GlancesAPI()
```

## Get plugins list

```python
gl.plugins()
```

## Get CPU stats

```python
gl.cpu
```

```python
gl.cpu['total']
```

## Get MEMORY stats

```python
gl.mem
```

```python
gl.mem['available']
```

Display it in a user friendly way:

```python
gl.auto_unit(gl.mem['available'])
```

## Get NETWORK stats

```python
gl.network
```

Get the list of networks interfaces:

```python
gl.network.keys()
```

Get stats for a specific network interface (`bytes_recv` is a rate, in bytes per second):

```python
iface = gl.network.keys()[0]
gl.network[iface]
```

```python
gl.network[iface]['bytes_recv']
```

## Additional information

Example for the LOAD plugin.

```python
gl.load
```

Get field descriptions and units:

```python
gl.load.fields
```

Get the thresholds configured in glances.conf:

```python
gl.load.limits
```

Get the current alert level of each watched field:

```python
gl.load.levels
```

## Keep collecting in the background

With `background=True` the Glances scheduler runs in a thread, so the history and the alerts fill up between reads.

```python
import time

bg = api.GlancesAPI(background=True, plugins=['cpu', 'mem'])
time.sleep(10)
bg.cpu.history()
```

```python
bg.alerts()
```

## Close the API

Or use it as a context manager: `with api.GlancesAPI() as gl: ...`

```python
bg.close()
gl.close()
```
